```python
#how many floors can we check with a given number of drops and glasses?
def max_floors(drops, glasses):

    # with t drops and g glasses
    table = [[0] * (glasses + 1)]

    for t in range(1, drops + 1):
        new_row = [0] * (glasses + 1)

        for g in range(1, glasses + 1):

            if g == 1:
                # Only one glass — we can only check one floor at a time
                new_row[g] = t
            else:
                previous = table[t - 1]

                if_broken   = previous[g - 1]
                if_survived = previous[g]

                new_row[g] = if_broken + 1 + if_survived

        table.append(new_row)

    return table[drops][glasses]


# how many drops do we need to check a given number of floors with a given number of glasses?
def min_drops(floors, glasses):
    drops = 0
    while max_floors(drops, glasses) < floors:
        drops += 1
    return drops

test_cases = [(10, 2), (100, 2), (100, 3), (1000, 3), (1000, 5)]

for floors, glasses in test_cases:
    result = min_drops(floors, glasses)
    print(f"{floors} floors, {glasses} glasses = {result} drops")
```
